# Численное дифференцирование по четырёхточечному шаблону

**Анализ дифференцируемости функции (Вариант 11)**

Функция: $f(x) = \frac{1}{3} + \cos(10 + 2,3^{|x|})$ на отрезке $x \in [-1; 2.5]$.
Данная функция содержит модуль $x$, то есть $|x|$. Производная модуля не существует в точке $x = 0$, которая входит в заданный промежуток $[-1; 2.5]$.
Проверим производную самой функции:
$$f'(x) = -\sin(10 + 2.3^{|x|}) \cdot 2.3^{|x|} \ln(2.3) \cdot \operatorname{sgn}(x)$$
Так как предел $\operatorname{sgn}(x)$ при $x \to 0^-$ равен $-1$, а при $x \to 0^+$ равен $1$, левая и правая производные в нуле не равны (а значение синуса в точке 0: $\sin(10 + 2.3^0) = \sin(11) \neq 0$).
Следовательно, функция не дифференцируема на заданном промежутке, поэтому выполню домашнее задание для следующего варианта (Вариант 12).

**Функция Вариант 12**

$f(x) = 2.5 \exp\left(\frac{x}{3} + 0.02\right) \cos(2x)$ на промежутке $x \in [-2; 2]$.
Данная функция является композицией гладких функций (экспоненты и косинуса) и дифференцируема на всем заданном отрезке.

**Вывод формул численного дифференцирования по 4 узлам**

Построим интерполяционный полином Ньютона для четырех узлов $x_0, x_1, x_2, x_3$ на равномерной сетке с шагом $h$ (то есть $x_i - x_j = (i - j)h$):

$$P_3(x) = f(x_0) + f[x_0, x_1](x - x_0) + f[x_0, x_1, x_2](x - x_0)(x - x_1) + f[x_0, x_1, x_2, x_3](x - x_0)(x - x_1)(x - x_2)$$

Обозначим значения функции в узлах: $f_i = f(x_i)$.
Запишем разделенные разности первого, второго и третьего порядков для равномерной сетки:
$$f[x_0, x_1] = \frac{f_1 - f_0}{h}$$
$$f[x_0, x_1, x_2] = \frac{f[x_1, x_2] - f[x_0, x_1]}{x_2 - x_0} = \frac{\frac{f_2 - f_1}{h} - \frac{f_1 - f_0}{h}}{2h} = \frac{f_2 - 2f_1 + f_0}{2h^2}$$
$$f[x_0, x_1, x_2, x_3] = \frac{f[x_1, x_2, x_3] - f[x_0, x_1, x_2]}{x_3 - x_0} = \frac{\frac{f_3 - 2f_2 + f_1}{2h^2} - \frac{f_2 - 2f_1 + f_0}{2h^2}}{3h} = \frac{f_3 - 3f_2 + 3f_1 - f_0}{6h^3}$$

Найдем производную полинома по $x$ (используем правило дифференцирования произведения):
$$f'(x) \approx P'_3(x) = f[x_0, x_1] + f[x_0, x_1, x_2] \cdot \big((x - x_1) + (x - x_0)\big) + f[x_0, x_1, x_2, x_3] \cdot \big((x - x_1)(x - x_2) + (x - x_0)(x - x_2) + (x - x_0)(x - x_1)\big)$$

Теперь подставим в эту формулу $x = x_0, x_1, x_2, x_3$ и получим значения производной в каждом узле:

1. **В узле $x_0$:**
Подставляем $x = x_0$, при этом $x_0 - x_1 = -h$, а $x_0 - x_2 = -2h$:
$$f'(x_0) \approx \frac{f_1 - f_0}{h} + \frac{f_2 - 2f_1 + f_0}{2h^2}(-h) + \frac{f_3 - 3f_2 + 3f_1 - f_0}{6h^3}(-h)(-2h) = \frac{f_1 - f_0}{h} - \frac{f_2 - 2f_1 + f_0}{2h} + \frac{f_3 - 3f_2 + 3f_1 - f_0}{3h}$$
Приводя к общему знаменателю $6h$, получаем:
$$f'(x_0) \approx \frac{6(f_1 - f_0) - 3(f_2 - 2f_1 + f_0) + 2(f_3 - 3f_2 + 3f_1 - f_0)}{6h} = \frac{-11f_0 + 18f_1 - 9f_2 + 2f_3}{6h}$$

2. **В узле $x_1$:**
Подставляем $x = x_1$, при этом $x_1 - x_0 = h$, а $x_1 - x_2 = -h$:
$$f'(x_1) \approx \frac{f_1 - f_0}{h} + \frac{f_2 - 2f_1 + f_0}{2h^2}(h) + \frac{f_3 - 3f_2 + 3f_1 - f_0}{6h^3}(h)(-h) = \frac{f_1 - f_0}{h} + \frac{f_2 - 2f_1 + f_0}{2h} - \frac{f_3 - 3f_2 + 3f_1 - f_0}{6h}$$
$$f'(x_1) \approx \frac{6(f_1 - f_0) + 3(f_2 - 2f_1 + f_0) - (f_3 - 3f_2 + 3f_1 - f_0)}{6h} = \frac{-2f_0 - 3f_1 + 6f_2 - f_3}{6h}$$

3. **В узле $x_2$:**
Подставляем $x = x_2$, при этом $x_2 - x_0 = 2h$, $x_2 - x_1 = h$:
$$f'(x_2) \approx \frac{f_1 - f_0}{h} + \frac{f_2 - 2f_1 + f_0}{2h^2}(h + 2h) + \frac{f_3 - 3f_2 + 3f_1 - f_0}{6h^3}(2h)(h) = \frac{f_1 - f_0}{h} + \frac{3(f_2 - 2f_1 + f_0)}{2h} + \frac{2(f_3 - 3f_2 + 3f_1 - f_0)}{6h}$$
$$f'(x_2) \approx \frac{6(f_1 - f_0) + 9(f_2 - 2f_1 + f_0) + 2(f_3 - 3f_2 + 3f_1 - f_0)}{6h} = \frac{f_0 - 6f_1 + 3f_2 + 2f_3}{6h}$$

4. **В узле $x_3$:**
Подставляем $x = x_3$, при этом $x_3 - x_0 = 3h$, $x_3 - x_1 = 2h$, $x_3 - x_2 = h$:
$$f'(x_3) \approx \frac{f_1 - f_0}{h} + \frac{f_2 - 2f_1 + f_0}{2h^2}(2h + 3h) + \frac{f_3 - 3f_2 + 3f_1 - f_0}{6h^3} \big((2h)(h) + (3h)(h) + (3h)(2h)\big)$$
Упрощая слагаемое: $2h^2 + 3h^2 + 6h^2 = 11h^2$, получаем:
$$f'(x_3) \approx \frac{f_1 - f_0}{h} + \frac{5(f_2 - 2f_1 + f_0)}{2h} + \frac{11(f_3 - 3f_2 + 3f_1 - f_0)}{6h}$$
$$f'(x_3) \approx \frac{6(f_1 - f_0) + 15(f_2 - 2f_1 + f_0) + 11(f_3 - 3f_2 + 3f_1 - f_0)}{6h} = \frac{-2f_0 + 9f_1 - 18f_2 + 11f_3}{6h}$$




```python
import numpy as np
import matplotlib.pyplot as plt

# Функция для варианта 12
def f(x):
    return 2.5 * np.exp(x / 3 + 0.02) * np.cos(2 * x)

# Аналитическая производная для проверки точности
def df_analytical(x):
    # f'(x) = 2.5 * exp(x/3 + 0.02) * (1/3) * cos(2x) - 2.5 * exp(x/3 + 0.02) * 2 * sin(2x)
    return 2.5 * np.exp(x / 3 + 0.02) * ((1/3) * np.cos(2 * x) - 2 * np.sin(2 * x))

# Заданный промежуток
a, b = -2, 2
n_nodes = 4

x_nodes = np.linspace(a, b, n_nodes)
h = x_nodes[1] - x_nodes[0]
f_nodes = f(x_nodes)

df_num = np.zeros(n_nodes)

# Формула для x0
df_num[0] = (-11 * f_nodes[0] + 18 * f_nodes[1] - 9 * f_nodes[2] + 2 * f_nodes[3]) / (6 * h)

# Формула для x1
df_num[1] = (-2 * f_nodes[0] - 3 * f_nodes[1] + 6 * f_nodes[2] - f_nodes[3]) / (6 * h)

# Формула для x2
df_num[2] = (f_nodes[0] - 6 * f_nodes[1] + 3 * f_nodes[2] + 2 * f_nodes[3]) / (6 * h)

# Формула для x3
df_num[3] = (-2 * f_nodes[0] + 9 * f_nodes[1] - 18 * f_nodes[2] + 11 * f_nodes[3]) / (6 * h)

print(f"Шаг сетки h = {h:.4f}\n")
print(f"{'Узел':>5} | {'x':>8} | {'Численная f(x)':>15} | {'Аналит. f(x)':>15} | {'Погрешность':>15}")
print("-" * 65)
for i in range(n_nodes):
    x_i = x_nodes[i]
    num_val = df_num[i]
    analyt_val = df_analytical(x_i)
    err = abs(num_val - analyt_val)
    print(f"x_{i:<3} | {x_i:8.4f} | {num_val:15.6f} | {analyt_val:15.6f} | {err:15.6e}")

x_dense = np.linspace(a, b, 200)
df_dense = df_analytical(x_dense)

plt.figure(figsize=(10, 6))

# Истинная производная
plt.plot(x_dense, df_dense, label="Аналитическая производная $f'(x)$", color='blue', linewidth=2)

plt.scatter(x_nodes, df_num, color='red', s=80, zorder=5, label='Численная производная (в 4 узлах)')

plt.title('Сравнение аналитической и численной производных (Вариант 12)', fontsize=14)
plt.xlabel('x', fontsize=12)
plt.ylabel("f'(x)", fontsize=12)
plt.grid(True, linestyle='--', alpha=0.7)
plt.legend(fontsize=12)
plt.tight_layout()
plt.show()
```
