```python
import numpy as np
import matplotlib.pyplot as plt
import random
```

```python
def f(x):
    return np.exp(x)

def F(x):
    return np.exp(x)

a = 1
b = 4
n = 10
```

```python
def divide_interval(a, b, n):
    return np.linspace(a, b, n + 1)

def random_points_integration(a, b, n):
    intervals = divide_interval(a, b, n)
    left_points = intervals[:-1]
    right_points = intervals[1:]
    random_points = [random.uniform(left, right) for left, right in zip(left_points, right_points)]

    fx = []
    I = 0
    for point in random_points:
        fx.append(f(point))
        I += f(point) * (b - a) / n

    return I, fx

I_random, fx_random = random_points_integration(a, b, n)

x = np.linspace(a, b, 1000)
y = f(x)
plt.plot(x, y, color="red", label="f(x)")
z = np.linspace(a, b, n + 1)
plt.bar(z[:-1], fx_random, width=(b - a) / n, align="edge", color="blue", alpha=0.7)
plt.xlabel("x")
plt.ylabel("f(x)")
plt.title(f"Метод случайных прямоушольников при n = {n}")
plt.grid()
plt.legend()
plt.show()
```
