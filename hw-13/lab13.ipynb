### Теоретическая справка

Согласно теореме о наилучшем приближении малого ранга (теорема Эккарта-Янга-Мирского), наилучшим приближением матрицы $A$ матрицей $B$ ранга не больше $k$ (по спектральной норме) является усеченное сингулярное разложение матрицы $A$:

$$A_k = \sum_{j=1}^{k} \sigma_j u^{(j)} (v^{(j)})^T$$

где $\sigma_j$ - сингулярные числа, отсортированные по убыванию ($\sigma_1 \ge \sigma_2 \ge \dots \ge \sigma_r > 0$), $u^{(j)}$ - левые сингулярные векторы (столбцы матрицы $U$), $v^{(j)}$ - правые сингулярные векторы (столбцы матрицы $V$).

При этом ошибка приближения (в спектральной норме) равна первому отброшенному сингулярному числу:
$$\|A - A_k\|_2 = \sigma_{k+1}$$

В данной задаче требуется найти приближение ранга $k = 3$. Для этого мы найдем сингулярное разложение матрицы $A$ и оставим только 3 наибольших сингулярных числа и соответствующие им векторы.

```python
import numpy as np
import pandas as pd

np.set_printoptions(precision=6, suppress=True)

VARIANT = 11
ROWS_IN_MATRIX = 15
EXPECTED_COLS = 14
CSV_PATH = "housing.csv"
K_RANK = 3

def read_housing_csv(path):
    for header_mode in ["infer", None]:
        df = pd.read_csv(path, header=header_mode)
        numeric_df = df.select_dtypes(include=[np.number])
        if numeric_df.shape[1] == EXPECTED_COLS:
            return numeric_df
    raise ValueError("Не удалось получить 14 числовых столбцов из файла housing.csv.")

# Чтение данных и построение матрицы
df = read_housing_csv(CSV_PATH)
data = df.to_numpy(dtype=float)

start_row = (VARIANT - 1) * ROWS_IN_MATRIX
end_row = VARIANT * ROWS_IN_MATRIX

A = data[start_row:end_row, :]

print(f"Исходная матрица A ({A.shape[0]}x{A.shape[1]}):")
print(A)

# Сингулярное разложение
# Используем встроенную функцию для получения точного SVD
U, S, Vt = np.linalg.svd(A, full_matrices=False)

print("\nСингулярные числа матрицы A:")
print(S)

# Построение приближения ранга 3
print(f"\nПостроение приближения ранга {K_RANK}")

# Оставляем только k первых компонент
U_k = U[:, :K_RANK]
S_k = np.diag(S[:K_RANK])
Vt_k = Vt[:K_RANK, :]

# Вычисляем приближенную матрицу A_k
A_k = U_k @ S_k @ Vt_k

print(f"\nПриближающая матрица A_{K_RANK} ({A_k.shape[0]}x{A_k.shape[1]}):")
print(A_k)

# Оценка ошибки
error_matrix = A - A_k
norm_error = np.linalg.norm(error_matrix, 2)

print("\nОценка ошибки")
print(f"Норма разности ||A - A_{K_RANK}||_2 : {norm_error:.6f}")
print(f"Ожидаемая ошибка (sigma_{K_RANK+1}) : {S[K_RANK]:.6f}")
```

В ходе работы для матрицы $A$ было построено наилучшее приближение ранга $k=3$ с помощью усеченного сингулярного разложения.

Согласно теореме о наилучшем приближении малого ранга, матрица $A_3$ была получена путем умножения матриц, составленных из 3 первых столбцов $U$, 3 первых строк $V^T$ и диагональной матрицы с 3 наибольшими сингулярными числами.

Проверка показала, что спектральная норма разности исходной матрицы и её приближения $\|A - A_3\|_2$ в точности равна четвертому сингулярному числу $\sigma_4$, что полностью подтверждает теоретическую оценку ошибки приближения.
