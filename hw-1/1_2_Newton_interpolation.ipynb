```python
def f(x):
    return np.sin(x)*np.cos(x)*x**2

def Newton_interpolation(x, y, t):

    n = len(x)

    coeffs = np.zeros((n, n))

    coeffs[:, 0] = y

    for j in range(1, n):
        for i in range(n - j):
            coeffs[i][j] = (coeffs[i + 1][j - 1] - coeffs[i][j - 1]) / (x[i + j] - x[i])

    result = coeffs[0, 0]
    omega = 1

    for j in range(1, n):
        omega *= (t - x[j - 1])
        result += coeffs[0, j] * omega

    return result

a, b = -1, 2
m = 5

xx = np.linspace(a, b, m)
yy = f(xx)

t = 1.333

int_p = Newton_interpolation(xx, yy, t)

plt.title("Интерполяционный полином в форме Ньютона")
plt.plot(xx, yy, color="red", marker="o")
plt.scatter(t, int_p, color="blue")
plt.show()
print("f(t) =", f(t))
print("Q(t) =", int_p)
```
