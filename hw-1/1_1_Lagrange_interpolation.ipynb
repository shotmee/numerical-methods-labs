```python
def f(x):
    return np.sin(x)*np.cos(x)*x**2  # функция

def Lagrange_interpolation(x, y, t):

    n = len(x)
    result = 0

    for i in range(n):
        L_i = 1
        for j in range(n):
            if i != j:
                L_i *= (t - x[j]) / (x[i] - x[j]) # L_i = w_i(t) / w_i(x_i)

        result += y[i] * L_i

    return result

a, b = -1, 2
m = 5

xx = np.linspace(a, b, m)


yy = f(xx)

t = 1.333

int_p = Lagrange_interpolation(xx, yy, t)

plt.title("Интерполяционный полином Лагранжа")
plt.plot(xx, yy, color="red", marker="o")
plt.scatter(t, int_p, color="blue")
plt.show()
print("f(t) =", f(t))
print("Q(t) =", int_p)
```
